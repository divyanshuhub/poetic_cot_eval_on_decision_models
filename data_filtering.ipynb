```python
import pandas as pd

input_path = "/home/gpuuser7/gpuuser7_a/Divyanshu/GSM8K/arc_decision_reasoning/outputs/vllm_cot_easy_test_full/ARC_easy_test.csv"
output_path = input_path.replace(".csv", "_complete_rows.csv")

df = pd.read_csv(input_path)

# Treat empty or whitespace-only cells as missing, then keep only complete rows.
df = df.replace(r"^\s*$", pd.NA, regex=True)
complete_rows = df.dropna(how="any")

complete_rows.to_csv(output_path, index=False)
print(f"Saved {len(complete_rows)} complete rows to {output_path}")
```
